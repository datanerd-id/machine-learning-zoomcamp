## Q6. Median value of horsepower

```python
median_before = df['horsepower'].median()
print('Median before filling:', median_before)

mode_hp = df['horsepower'].mode()[0]
print('Most frequent value (mode):', mode_hp)

horsepower_filled = df['horsepower'].fillna(mode_hp)
median_after = horsepower_filled.median()

print('Median after filling:', median_after)
```

**Answer:** Yes, it decreased (from 254.0 to 252.0)

## Q7. Sum of weights

```python
# Select Asia cars, columns vehicle_weight and model_year, first 7 rows
X_df = df[df['origin'] == 'Asia'][['vehicle_weight', 'model_year']].head(7)
X_df
```

```python
# Get the underlying NumPy array
X = X_df.values
X
```

```python
# Compute the matrix product of X transposed and X
XTX = X.T.dot(X)
XTX
```

```python
# Invert the matrix XTX
XTX_inv = np.linalg.inv(XTX)
XTX_inv
```

```python
# 7: Create y
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
y
```

```python
# Compute the weights using the normal equation
w = XTX_inv.dot(X.T).dot(y)
w
```

```python
# Sum of the weights
w.sum()
```
